```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
# Load the data
mukuna = pd.read_csv('/content/Road Accident Data.csv')
# Extract the independent variable (X) and dependent variable (y)
X = mukuna[['Number_of_Vehicles']]
y = mukuna['Number_of_Casualties']
# Create and fit a linear regression model
regressor = LinearRegression()
regressor.fit(X, y)
# Plot the data points
mukuna.plot(x='Number_of_Vehicles', y='Number_of_Casualties', style='o', label='Data Points')
# Overlay the regression line
plt.plot(X, regressor.predict(X), color='red', linewidth=2, label='Regression Line')
plt.xlabel('Number_of_Vehicles')
plt.ylabel('Number_of_Casualties')
plt.legend()
plt.show()
```
